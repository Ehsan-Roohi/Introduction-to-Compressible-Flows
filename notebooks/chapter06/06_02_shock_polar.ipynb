# Shock-Polar Diagram Across Upstream Mach Number

**Book:** *Introduction to Compressible Flows*<br>
**Chapter 6:** Oblique Shock Waves and Expansion Waves<br>
**Concepts:** shock polar · velocity-space construction · Mach-number dependence

## Learning objectives

- Construct a family of shock polars for air at a prescribed temperature.
- Visualize how the admissible post-shock velocity locus changes with Mach number.
- Export a publication-quality polar plot for further interpretation.

## How to use this notebook

Run the code cell; it writes `R_phi_polar_90_180.png` in the current working directory.

> **Model scope.** The plot uses a perfect-gas model with fixed gamma and temperature and is intended as a geometric teaching aid.


```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------
# Physical constants
# ------------------------
gamma = 1.4
T = 300.0               # K
Ru = 8314.0
Mw_air = 28.96
R_air = Ru / Mw_air

# Speed of sound
a = np.sqrt(gamma * R_air * T)

# Mach numbers (1.2 to 10 with step 1)
M_list = np.arange(1.2, 10.1, 1.0)

# Phi from 90 to 180 degrees
phi = np.linspace(np.pi/2, np.pi, 600)

# ------------------------
# Polar plot
# ------------------------
plt.figure(figsize=(8,8))
ax = plt.subplot(111, projection='polar')

for M in M_list:
    V1 = M * a
    r = (2 * V1 / (gamma + 1)) * (np.cos(phi)/M**2 - np.cos(phi))
    ax.plot(phi, r, linewidth=1.6, label=f"$M_1={M:.1f}$")

# Show only upper half of the circle
ax.set_thetalim(0, np.pi)

ax.set_title(r"$R(\phi)\ \mathrm{vs}\ \phi\quad (\phi\in[90^\circ,180^\circ])$",
             fontsize=13)

ax.set_theta_zero_location("E")
ax.set_theta_direction(1)

ax.legend(loc="upper right", bbox_to_anchor=(1.28, 1.05), fontsize=13)
ax.grid(True)

# ------------------------
# Save figure (no display)
# ------------------------
plt.savefig("R_phi_polar_90_180.png", dpi=300, bbox_inches="tight")
plt.close()
```
