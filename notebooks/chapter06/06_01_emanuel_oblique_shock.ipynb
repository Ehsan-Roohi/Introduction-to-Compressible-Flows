# Weak and Strong Oblique-Shock Angles by Emanuel's Explicit Method

**Book:** *Introduction to Compressible Flows*<br>
**Chapter 6:** Oblique Shock Waves and Expansion Waves<br>
**Concepts:** theta-beta-Mach relation · weak and strong shock branches · shock detachment

## Learning objectives

- Evaluate the explicit cubic solution for the shock angle.
- Distinguish weak and strong attached-shock branches.
- Detect inputs beyond the attached-shock limit.

## How to use this notebook

Edit the worked-example Mach number, turning angle, or heat-capacity ratio in the final lines of the code cell and rerun it.

> **Model scope.** The method assumes a calorically perfect gas and an attached, two-dimensional oblique shock.


```python
import numpy as np

def obliqueshock_emanuel(M1, theta_deg, gamma=1.4):
    """
    Computes the oblique shock angles (Weak and Strong) using Emanuel's explicit analytical method.
    Reference: Emanuel, G., 'Gasdynamics: Theory and Applications', AIAA Education Series.
    """
    theta = np.radians(theta_deg)
    m2_minus_1 = M1**2 - 1.0

    # Coefficients for the cubic equation solution
    term1 = 1.0 + 0.5 * (gamma - 1.0) * M1**2
    term2 = 1.0 + 0.5 * (gamma + 1.0) * M1**2
    tan2_theta = np.tan(theta)**2

    # Check for the detachment limit (discriminant of the cubic equation)
    lambda_sq = m2_minus_1**2 - 3.0 * term1 * term2 * tan2_theta
    if lambda_sq < 0:
        return None, None, 'Detached shock: deflection angle theta exceeds theta_max for this Mach number.'

    lam = np.sqrt(lambda_sq)

    # Calculating the chi parameter for the analytical solution
    chi_num = (m2_minus_1**3 - 9.0 * term1 *
               (1.0 + 0.5 * (gamma - 1.0) * M1**2 + 0.25 * (gamma + 1.0) * M1**4) * tan2_theta)
    chi_den = lam**3
    chi = chi_num / chi_den

    # Clipping chi to avoid numerical issues at the physical limit
    chi = np.clip(chi, -1.0, 1.0)

    denom = 3.0 * term1 * np.tan(theta)

    # Weak Shock Branch calculation
    arg_weak = (4.0 * np.pi + np.arccos(chi)) / 3.0
    tan_beta_weak = (m2_minus_1 + 2.0 * lam * np.cos(arg_weak)) / denom
    beta_weak_deg = np.degrees(np.arctan(tan_beta_weak))

    # Strong Shock Branch calculation
    arg_strong = np.arccos(chi) / 3.0
    tan_beta_strong = (m2_minus_1 + 2.0 * lam * np.cos(arg_strong)) / denom
    beta_strong_deg = np.degrees(np.arctan(tan_beta_strong))

    return beta_weak_deg, beta_strong_deg, 'Success'

def main():
    print("=== Oblique Shock Angle Calculator (Emanuel's Explicit Method) ===")

    try:
        # User input for Gamma (Ratio of Specific Heats)
        gamma_input = input("Enter ratio of specific heats (gamma) [Default 1.4]: ").strip()
        gamma = float(gamma_input) if gamma_input else 1.4

        # User input for Mach Number and Deflection Angle
        M1 = float(input("Enter upstream Mach number (M1): "))
        theta_deg = float(input("Enter flow deflection angle (theta) in degrees: "))

        # Function call and result retrieval
        beta_w, beta_s, msg = obliqueshock_emanuel(M1, theta_deg, gamma)

        print("\n--- Calculation Results ---")
        if beta_w is not None:
            print(f"Input Mach (M1): {M1}")
            print(f"Deflection Angle (theta): {theta_deg}°")
            print(f"Specific Heat Ratio (gamma): {gamma}")
            print(f"Weak Shock Angle (beta_w): {beta_w:.4f}°")
            print(f"Strong Shock Angle (beta_s): {beta_s:.4f}°")
        else:
            print(f"Warning/Error: {msg}")

    except ValueError:
        print("\nError: Please enter valid numerical values!")





# Reproducible worked example
M1_example = 2.0
theta_example_deg = 10.0
beta_w, beta_s, message = obliqueshock_emanuel(M1_example, theta_example_deg)
print(f"M1 = {M1_example:.2f}, theta = {theta_example_deg:.2f} deg")
print(f"weak beta = {beta_w:.4f} deg, strong beta = {beta_s:.4f} deg")
print(message)
```

## Interpretation and verification checklist

- The weak-shock angle must lie between the Mach angle and the strong-shock angle.
- As the turning angle approaches zero, the weak branch approaches the Mach angle.
- A detached-shock message is a physical-domain result, not a numerical failure.
